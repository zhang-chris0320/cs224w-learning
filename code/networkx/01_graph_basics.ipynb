# Graph Basics with NetworkX

A minimal graph example for the Transformer-as-Node research idea.

```python
import networkx as nx

G = nx.Graph()

G.add_nodes_from([
    "Transformer-A",
    "Transformer-B",
    "Transformer-C"
])

G.add_edges_from([
    ("Transformer-A", "Transformer-B"),
    ("Transformer-B", "Transformer-C"),
])

print(G.nodes())
print(G.edges())
```
